# **2. Limpieza de datos**

Realiza las transformaciones necesarias para obtener un DataFrame consistente.
Debes:
    • Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.
    • Detectar y elimina los registros duplicados cuando corresponda.
    • Convertir las columnas que tengan tipos de datos incorrectos.
    • Identificar al menos dos variables categóricas que presenten inconsistencias y estandarízalas.
    • Después de realizar la limpieza, demuestra mediante código que las transformaciones fueron aplicadas correctamente.

```python
### Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.
display(df[df["Animal type"].isnull()].head(20))
df = df.dropna(subset=["Animal type"])
df["Animal type"].isnull().sum()
```

```python
df["Country"].isnull().sum()
#se borran los 12 valores
```

```python
# 3. Verificando valores nulos en columna:   Peso en Kg
display(df[df["Weight kg"].isnull()].head(27))

df["Animal type"] = (
    df["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)
animales = {
    "Red Squirrell" : "Red Squirrel",
    "Red Squirel" : "Red Squirrel",
    "European Bison™" : "European Bison",
    "European Bisson" : "European Bison",
    "European Buster" : "European Bison",
    "Ledgehod" :"Hedgehog",
    "Wedgehod" : "Hedgehog",
    "Lynx?" : "Lynx"
}
df["Animal type"] = df["Animal type"].replace(animales)
peso_ardillas = df[df["Animal type"] == "Red Squirrel"]
peso_erizos = df[df["Animal type"] == "Hedgehog"]
df["Weight kg"] = df["Weight kg"].fillna(peso_ardillas["Weight kg"].median())
df["Weight kg"] = df["Weight kg"].fillna(peso_erizos["Weight kg"].median())
df["Weight kg"].isnull().sum()
```

```python
display(df[df["Body Length cm"].isnull()].head(27))
mediana_long_ardillas = df[df["Animal type"] == "Red Squirrel"]["Body Length cm"].median()
mediana_long_erizos = df[df["Animal type"] == "Hedgehog"]["Body Length cm"].median()
df["Body Length cm"] = df["Body Length cm"].fillna(mediana_long_ardillas)
df["Body Length cm"] = df["Body Length cm"].fillna(mediana_long_erizos)
df["Body Length cm"].isnull().sum()
```

```python
display(df[df["Gender"].isnull()].head(19))

df['Gender']= df['Gender'].fillna("not determined")
df["Gender"].isnull().sum()
df["Gender"].value_counts()
```

```python
display(df[df["Latitude"].isnull()].head(100))

df["Country"] = (
    df["Country"]
    .str.strip()
    .str.lower()
    .str.title()
)
Paises = {
    "Pl":"Poland",
    "Hu" :"Hungary",
    "Hungry" :"Hungary",
    "Cz": "Czech Republic",
    "Czech": "Czech Republic",
    "Cc":"Czech Republic",
    "De" :"Germany "
}
df["Country"]= df["Country"].replace(Paises)
df["Country"].value_counts()


df["Latitude"] = df["Latitude"].fillna(df.groupby("Country")["Latitude"].transform("mean"))
df["Longitude"] = df["Longitude"].fillna(df.groupby("Country")["Longitude"].transform("mean"))
df["Latitude"].isnull().sum()
```

```python
display(df[df["Animal name"].isnull()].head(959))
df["Animal name"] = df["Animal name"].fillna("No especificado")
df["Animal name"].isnull().sum()
```

```python
if 'Animal code' in df.columns:
    df = df.drop(columns=['Animal code'])
```
